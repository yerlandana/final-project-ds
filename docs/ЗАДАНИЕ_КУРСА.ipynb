### 📌 Что такое итоговый проект?

Итоговый проект — это ваш **первый реальный проект для портфолио**. Вы берёте задачу, данные, строите модель и делаете рабочий продукт, который можно показать работодателю.

**Требования:**
- Рабочая модель с измеримыми метриками
- FastAPI интерфейс (пользователь может взаимодействовать)
- Код на GitHub с README
- 5-минутная презентация

**Дедлайн:** 08.09.2026

## 🛤️ Два трека

У вас два направления на выбор. Можно выбрать один **или** сделать гибридный проект (ML + AI).

---

### Трек 1: 📊 Machine Learning

**Суть:** Берём структурированные данные (таблицы), строим модель, предсказываем целевую переменную.

**Что используем из курса:**
- Pandas, NumPy (недели 5-7)
- EDA и визуализация (неделя 10)
- Метрики, train/test split (неделя 13)
- Регрессия или классификация (недели 14-16)
- Feature Engineering, Pipelines (неделя 17)
- XGBoost, Random Forest (неделя 16)

---

### Трек 2: 🤖 AI / NLP

**Суть:** Работаем с текстом, языковыми моделями, API. Строим интеллектуальное приложение.

**Что используем из курса:**
- LLM, трансформеры, эмбеддинги
- Prompt Engineering, LLM API
- RAG, ChromaDB
- Fine-tuning
- AI-агенты, LangChain, инструменты

## 🔥 Гибридные проекты: ML + AI

Самые интересные проекты — те, которые **комбинируют оба трека**. Это выделит вас в портфолио.

### Идея 1: 🏥 Медицинский ассистент
- **ML часть:** модель предсказывает вероятность заболевания (диабет, сердце) по анализам
- **AI часть:** LLM объясняет результат пациенту понятным языком, даёт рекомендации


### Идея 2: 🏠 Умный оценщик недвижимости
- **ML часть:** XGBoost предсказывает цену квартиры по параметрам
- **AI часть:** LLM анализирует описание объявления, извлекает features, генерирует отчёт


### Идея 3: 📊 Аналитик отзывов для бизнеса
- **ML часть:** классификация тональности отзывов (позитив/негатив/нейтрал)
- **AI часть:** LLM суммаризирует основные жалобы и предлагает улучшения


### Идея 4: 💼 HR-ассистент (скрининг резюме)
- **ML часть:** модель ранжирует кандидатов по фичам (опыт, навыки, образование)
- **AI часть:** LLM читает текст резюме, извлекает структурированные данные, генерирует summary


### Идея 5: 🎓 Умный тьютор с аналитикой
- **ML часть:** модель предсказывает слабые темы ученика по истории ответов
- **AI часть:** LLM генерирует персонализированные вопросы и объяснения


### Идея 6: 🛒 Рекомендательная система с объяснениями
- **ML часть:** collaborative filtering или content-based рекомендации
- **AI часть:** LLM объясняет почему рекомендован этот товар/фильм


### Почему гибрид — это круто?
- Показывает владение **обоими** направлениями
- Ближе к реальным продуктам (ML модель + человеко-понятный интерфейс)
- Выделяет портфолио среди других джуниоров

## 🗺️ Roadmap проекта: пошаговый план

### День 1 (Сегодня → до следующего урока)

```
Шаг 1: Выбор темы (30 мин)
├── Выбрать трек (ML / AI / Гибрид)
├── Выбрать конкретную тему
└── Найти датасет

Шаг 2: Разведка данных — EDA (1-2 часа)
├── Загрузить данные
├── df.info(), df.describe(), df.shape
├── Пропуски, дубликаты
├── Визуализация таргета и ключевых фичей
└── Первые выводы

Шаг 3: Baseline модель (1-2 часа)
├── Минимальная предобработка
├── Train/test split
├── Простая модель
└── Замерить метрики → записать
```

### День 2-3 (Между уроками)

```
Шаг 4: Улучшение модели (2-3 часа)
├── Feature engineering
├── Попробовать 2-3 модели
├── Подбор гиперпараметров
├── Сравнительная таблица метрик
└── Выбрать лучшую модель


### День 4 (Перед уроком 2)

```
Шаг 6: Оформление (1-2 часа)
├── GitHub репозиторий
├── README.md (описание, как запустить, результаты)
├── requirements.txt
└── Подготовить 5-минутную презентацию
```


---
## 📁 Структура проекта

```
